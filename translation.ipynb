```python
from googletrans import Translator

# Initialize Translator
translator = Translator()

# Function to translate text
def translate_text(text, target_language):
    try:
        translated = translator.translate(text, dest=target_language)
        return translated.text
    except Exception as e:
        return f"Translation failed: {e}"

# Take input from the user
text = input("Enter text to translate: ")
target_language = input("Enter target language (e.g., fr for French, es for Spanish, hi for Hindi): ")

# Perform translation
if text and target_language:
    translated_text = translate_text(text, target_language)
    print("\n✅ **Translated Text:**\n", translated_text)
else:
    print("❌ Please provide both text and target language!")
```
