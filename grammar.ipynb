```python


import google.generativeai as genai
import ipywidgets as widgets
from IPython.display import display, HTML

# Configure Gemini API (Make sure you have a valid API key)
genai.configure(api_key="AIzaSyB4l0hpYkq9iE77vj2yqxhkLO4uu75KbFQ")

# Function to correct grammar using Google Gemini
def correct_grammar(text):
    model = genai.GenerativeModel("gemini-pro")
    response = model.generate_content(f"Correct this sentence: {text}")
    return response.text

# UI Elements
text_input = widgets.Textarea(
    placeholder="Enter text here...",
    layout=widgets.Layout(width="100%", height="120px")
)

correct_button = widgets.Button(description="Correct Grammar", button_style='info')

output_area = widgets.Output()

# Button Click Event
def on_button_click(b):
    with output_area:
        output_area.clear_output()
        text = text_input.value.strip()
        if not text:
            print("❌ Please enter some text!")
        else:
            corrected_text = correct_grammar(text)
            print("✅ **Corrected Text:**\n", corrected_text)

correct_button.on_click(on_button_click)

# Display UI in Jupyter Notebook
display(text_input, correct_button, output_area)
```
